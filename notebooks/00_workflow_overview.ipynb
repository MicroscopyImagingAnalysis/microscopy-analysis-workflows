## 1. Define the analysis contract

Each image becomes a manifest row. Segmentation adds labelled arrays; feature
extraction adds object tables; representations add graph or latent tables. A
stable object identifier links them throughout the workflow.

```python
import pandas as pd

workflow = pd.DataFrame(
    [
        ("image", "image array + metadata", "segmentation"),
        ("segmentation", "label image + QC", "measurement"),
        ("measurement", "object feature table", "representation"),
        ("representation", "graph/latent feature table", "analysis"),
        ("analysis", "statistics + figures", "interpretation"),
    ],
    columns=["stage", "product", "next_stage"],
)
workflow
```
